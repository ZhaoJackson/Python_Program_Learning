### Class: a blueprint of creating object
- attributes (what it has)
- methods (functions that operate)
1. A class defines how an object should behave and what data it should store.
2. Objects created from a class are called instances.
3. Classes help organize code, make it reusable, and support OOP principles like encapsulation, inheritance, and polymorphism

```python
# step 1: create a class website for users
class User:
    pass

# step 2: create an object from the class User
user_1 = User()

# step 3: create an attribute for the class User
user_1.id = "001"
user_1.username = "angela"

# step 4: access the attributes of the object
print(user_1.id)
print(user_1.username)
```
